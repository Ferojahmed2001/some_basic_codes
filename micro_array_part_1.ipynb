##### Now we  can decide what type of dtring analysis is needed

```python
#### run this if the longest gene character has ///


genes = series_mat_official['genes'].str.split("///")

official_gene_symbol = []

for i in genes:
    if isinstance(i, float):
        official_gene_symbol.append(i)
    else:
        official_gene_symbol.append(i[0])
```

```python
#### run this if the longest gene character has //


genes = series_mat_official['genes'].str.split("//")

official_gene_symbol = []

for i in genes:
    if isinstance(i, float):
        official_gene_symbol.append(i)
    elif '---' in i[0]:
        official_gene_symbol.append(pd.NA)
    elif len(i)  > 1:
        official_gene_symbol.append(i[1])
    else:
        official_gene_symbol.append(i[0])


```

```python
print(len(official_gene_symbol ))
series_mat_official['genes']=official_gene_symbol 
series_mat_official['genes'] =  series_mat_official["genes"].str.strip()
```

```python
series_mat_official.dropna(subset="genes", inplace=True)
series_mat_official.shape
```

```python
duplicate_genes = series_mat_official["genes"][series_mat_official.duplicated(subset=("genes"))].unique()
print(duplicate_genes[0:11])
print(len(duplicate_genes))
```

```python
unique_data = series_mat_official.drop_duplicates(subset='genes', keep= False)
unique_data.shape
```

```python
data3 = pd.DataFrame(index= duplicate_genes, columns= series_mat_official.columns)

for i in duplicate_genes:
    gene = series_mat_official[series_mat_official['genes']==i].set_index("genes")
    value = gene.mean(axis=0).to_list()
    value.insert(0, i)
    data3.loc[i,:] = value
```

```python
data4  = pd.concat((unique_data,data3),axis = 0,ignore_index=True)
data4.shape
```

```python
data4.to_csv("GSE48780_clean.csv",index=False)
```
