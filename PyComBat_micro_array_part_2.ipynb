```python
from functools import reduce

data = [dataset_1_transform,dataset_2_transform,
        dataset_3_transform,dataset_4_transform]

result = reduce(lambda left, right: pd.merge(left, right, left_index=True, right_index=True, how='inner'), data)

data = pd.DataFrame(result)

print(result.shape)
```

```python
from combat.pycombat import pycombat
```

```python
fig,ax =  plt.subplots(figsize = (20,20))

batch = []
datasets = [dataset_1_common,dataset_2_common,
            dataset_3_common,dataset_4_common]
for j in range(len(datasets)):
    batch.extend([j for _ in range(len(datasets[j].columns))])


df_corrected = pycombat(result,batch)


plt.boxplot(df_corrected)
plt.show()
```
