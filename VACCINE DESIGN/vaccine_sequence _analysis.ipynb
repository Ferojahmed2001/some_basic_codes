```python
from Bio import AlignIO
```

```python
alignment_vp1 = AlignIO.read("vp1_aligned.fasta", "fasta")
print(alignment_vp1)
```

```python
alignment_vp1.get_alignment_length()
```

```python

conserve_sequence = ""
for residue in range(0,alignment_vp1.get_alignment_length()):
    unique_chars = ""
    for i in alignment_vp1[:,residue]:
        if i in unique_chars:
            pass
        else:
            unique_chars+=i
    char_dict = dict()
    for j in unique_chars:

        char_dict.update({f"{j}":alignment_vp1[:,residue].count(j)})

    char_dict
    dict_value= list(char_dict.values())
    for keys, values in char_dict.items():
        
        if keys != "-":
        
            if values == max(dict_value):

                conserve_sequence+= keys
                break  ### if there is more than two residue occurs most then it will take the first one alphabetically
        else:
            dict_value.remove(max(dict_value))
            
            if values == max(dict_value):
                conserve_sequence+= keys
                break


conserve_sequence
```

```python
conserved = ""
for i in range(alignment_vp1.get_alignment_length()):
    column = alignment_vp1[:, i]
    if column.count(column[0]) == len(column) and column[0] != "-":
        conserved += column[0]
    else:
        conserved += "-"

print("Conserved pattern:\n", conserved)
```
