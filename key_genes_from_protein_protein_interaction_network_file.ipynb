```python
import os
import pandas as pd
```

```python

directory_path = '/home/kaderi/Feroj'
concatenated_df = pd.DataFrame()


for filename in os.listdir(directory_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(directory_path, filename)
        df = pd.read_csv(file_path, skiprows=1, usecols=[1])
        df.columns = [os.path.splitext(filename)[0]]
        concatenated_df = pd.concat([concatenated_df, df], axis=1)


df = concatenated_df


all_values = df.values.flatten()
value_series = pd.Series(all_values)

unique_values = value_series.drop_duplicates()
value_counts = value_series.value_counts()
value_counts_df = value_counts.reset_index()
value_counts_df.columns = ['Value Name', 'Occurrence']
top_10_values = value_counts_df.head(15)
df_combined2 = pd.concat([df,pd.DataFrame(['*']*12),top_10_values], ignore_index=False, axis =1)

df_combined2.to_csv("/home/kaderi/Feroj/T2DM&TNBC/results_section/hub_genes.csv", index=False)
```
